- There are 8 categorical features: X0, X1, X2, X3, X4, X5, X6, and X8.
- There are 368 binary features.

- Several features show noticeable correlation with the target.

- Categorical features will require encoding during preprocessing.
- The large number of features motivates feature shrinkage/dimensionality reduction.

# 2. Data Preprocessing

### 2.1 Removing Constant Features

```python
X_train = X_train.drop(columns=constant_features, errors='ignore')
X_val = X_val.drop(columns=constant_features, errors='ignore')

print("Training Shape",X_train.shape)
print("Validation Shape", X_val.shape)
```

### 2.2 Separating Categorical and Numerical Features

```python
categorical_features = [
    col for col in X_train.columns
    if X_train[col].dtype == "object"
]

numerical_features = [
    col for col in X_train.columns
    if X_train[col].dtype != "object"
]

print("Number of categorical features:", len(categorical_features))
print("Number of numerical features:", len(numerical_features))

print("\nCategorical features:")
print(categorical_features)
```

### 2.3 One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_train_cat = encoder.fit_transform(
    X_train[categorical_features]
)

X_val_cat = encoder.transform(
    X_val[categorical_features]
)

print("Encoded training shape:", X_train_cat.shape)
print("Encoded validation shape:", X_val_cat.shape)
```

```python
X_train_cat
```

### 2.4 Keeping Binary Features

```python
binary_features = [
    col for col in X_train.columns
    if col not in categorical_features
]

X_train_binary = X_train[binary_features]
X_val_binary = X_val[binary_features]
```

```python
X_train_binary
```

### 2.5 Combining Encoded and Binary Features

```python
X_train_processed = np.hstack([
    X_train_binary.values,
    X_train_cat
])

X_val_processed = np.hstack([
    X_val_binary.values,
    X_val_cat
])
```

```python
print("Processed training shape:", X_train_processed.shape)
print("Processed validation shape:", X_val_processed.shape)
```

```python
X_train_processed
```

### 2.6 Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_processed)
X_val_scaled = scaler.transform(X_val_processed)
```

```python
X_train_scaled
```

### 2.7 Lasso Feature Selection

```python
from sklearn.linear_model import LassoCV

lasso = LassoCV(
    cv=5,
    random_state=42,
    max_iter=10000,
    n_jobs=-1
)

lasso.fit(X_train_scaled, y_train)

print("Best alpha:", lasso.alpha_)
```

```python
lasso_selected_mask = lasso.coef_ != 0

print("Original number of features:", X_train_scaled.shape[1])
print("Selected features:", lasso_selected_mask.sum())
print("Removed features:", (~lasso_selected_mask).sum())
```

```python
processed_feature_names = (
    numerical_features
    + encoder.get_feature_names_out(categorical_features).tolist()
)

print("Number of feature names:", len(processed_feature_names))
```

```python
selected_feature_names = [
    feature
    for feature, selected in zip(
        processed_feature_names,
        lasso_selected_mask
    )
    if selected
]

print("Number of selected feature names:", len(selected_feature_names))
print("\nFirst selected features:")
print(selected_feature_names[:20])
```

```python
X_train_lasso = X_train_scaled[:, lasso_selected_mask]
X_val_lasso = X_val_scaled[:, lasso_selected_mask]

print("Lasso training shape:", X_train_lasso.shape)
print("Lasso validation shape:", X_val_lasso.shape)
```

### 2.8 Principal Component Analysis (PCA)

```python
from sklearn.decomposition import PCA

pca = PCA(
    n_components=0.95,
    random_state=42
)

X_train_pca = pca.fit_transform(X_train_scaled)
X_val_pca = pca.transform(X_val_scaled)

print("Original number of features:", X_train_scaled.shape[1])
print("Number of PCA components:", pca.n_components_)

print("PCA training shape:", X_train_pca.shape)
print("PCA validation shape:", X_val_pca.shape)

print(
    "Explained variance:",
    pca.explained_variance_ratio_.sum()
)
```

```python
print("Before PCA:", X_train_scaled.shape)
print("After PCA:", X_train_pca.shape)
```

```python
cumulative_variance = np.cumsum(
    pca.explained_variance_ratio_
)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Cumulative Explained Variance")

plt.grid(alpha=0.3)
plt.show()
```

## 3. Modeling & Tuning

### 3.1 Importing Evaluation Metrics

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
```

### 3.2 Defining a Function to Evaluate Models

```python
def evaluate_model(model, X_train, y_train, X_val, y_val):
    # Train the model
    model.fit(X_train, y_train)

    # Predictions
    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_val)

    # Training metrics
    train_r2 = r2_score(y_train, y_train_pred)
    train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
    train_mae = mean_absolute_error(y_train, y_train_pred)

    # Validation metrics
    val_r2 = r2_score(y_val, y_val_pred)
    val_rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))
    val_mae = mean_absolute_error(y_val, y_val_pred)

    return {
        "Train R²": train_r2,
        "Train RMSE": train_rmse,
        "Train MAE": train_mae,
        "Validation R²": val_r2,
        "Validation RMSE": val_rmse,
        "Validation MAE": val_mae
    }
```

### 3.3 Baseline: All 549 Features

```python
ridge_baseline = Ridge(alpha=1.0)

baseline_results = evaluate_model(
    ridge_baseline,
    X_train_scaled,
    y_train,
    X_val_scaled,
    y_val
)

print("Baseline Ridge Results")
print("-" * 30)

for metric, value in baseline_results.items():
    print(f"{metric}: {value:.6f}")
```

### 3.4 Lasso

```python
ridge_lasso = Ridge(alpha=1.0)

lasso_results = evaluate_model(
    ridge_lasso,
    X_train_lasso,
    y_train,
    X_val_lasso,
    y_val
)

print("Lasso-Selected Ridge Results")
print("-" * 30)

for metric, value in lasso_results.items():
    print(f"{metric}: {value:.6f}")
```

### 3.5 PCA

```python
ridge_pca = Ridge(alpha=1.0)

pca_results = evaluate_model(
    ridge_pca,
    X_train_pca,
    y_train,
    X_val_pca,
    y_val
)

print("PCA Ridge Results")
print("-" * 30)

for metric, value in pca_results.items():
    print(f"{metric}: {value:.6f}")
```

```python
pca_thresholds = [0.80, 0.85, 0.90, 0.95, 0.99]

pca_results_list = []
```
